| Topic                 | What is it?                                              | Why?                                                   | Problem Solved                               | When Used                                   | Example                                   |
| --------------------- | -------------------------------------------------------- | ------------------------------------------------------ | -------------------------------------------- | ------------------------------------------- | ----------------------------------------- |
| **Embeddings**        | Convert text into numerical vectors representing meaning | Computers understand numbers, not language             | Enables understanding of semantic similarity | Before storing/searching text               | "Car" ≈ "Automobile"                      |
| **Vector Database**   | Database optimized for storing and searching embeddings  | Fast similarity search over millions of vectors        | Makes semantic search scalable               | Large document or AI systems                | Pinecone, Chroma, pgvector                |
| **Document Search**   | Search documents by meaning instead of exact words       | Finds relevant information even with different wording | Avoids keyword-only limitations              | PDFs, manuals, policies, research           | Search "leave" and find "vacation policy" |
| **AI Knowledge Base** | Collection of private documents used by AI               | Gives LLM access to company-specific knowledge         | Answers questions using internal data        | Enterprise chatbots, HR assistants, support | Company policy documents                  |

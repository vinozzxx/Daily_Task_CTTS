React - 
A JavaScript library used to build reusable and interactive user interfaces.

Component - 
A reusable piece of UI that encapsulates its own structure and behavior.

Props - 
Read-only data passed from a parent component to a child component.

State - 
Data managed by a component that can change over time and trigger UI updates.

Hook - 
A special React function that lets functional components use state and other React features.

Tailwind CSS - 
A utility-first CSS framework for building custom user interfaces quickly.

Responsive UI - 
A user interface that automatically adapts to different screen sizes and devices.

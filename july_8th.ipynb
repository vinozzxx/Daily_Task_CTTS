React - 
A JavaScript library used to build reusable and interactive user interfaces.

Component - 
A reusable piece of UI that encapsulates its own structure and behavior.

Props - 
Read-only data passed from a parent component to a child component.

State - 
Data managed by a component that can change over time and trigger UI updates.

Hook - 
A special React function that lets functional components use state and other React features.

Tailwind CSS - 
A utility-first CSS framework for building custom user interfaces quickly.

Responsive UI - 
A user interface that automatically adapts to different screen sizes and devices.

Next.js - 
A React framework for building production-ready full-stack web applications with routing, server rendering, API routes, and performance optimizations.

App Router -
A file-system based routing system where folders and files automatically become URLs.

Server Component -
A React component that runs on the server and sends rendered HTML to the client.

Client Component -
A React component that runs in the browser and supports interactivity using state, events, and hooks.

API Route - 
A backend endpoint built directly inside a Next.js application to handle HTTP requests.

Environment Variable - 
A configuration value used to store secrets and settings outside the source code.

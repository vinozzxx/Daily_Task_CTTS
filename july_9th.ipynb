| Concept              | What (Technical)                                                                                              | Why                                                               | When to Use                                                   | Memory Trick               |
| -------------------- | ------------------------------------------------------------------------------------------------------------- | ----------------------------------------------------------------- | ------------------------------------------------------------- | -------------------------- |
| **State**            | Component data that can **change over time** and trigger a UI re-render.                                      | To display dynamic information.                                   | Any data that changes (counter, input, user, etc.).           | 📦 **Data**                |
| **State Management** | The process of **storing, updating, and sharing state** across components.                                    | To keep the UI synchronized and consistent.                       | When multiple components need the same data.                  | 🌐 **Data Manager**        |
| **Context API**      | React's built-in mechanism for **sharing global state** without prop drilling.                                | To avoid passing props through many intermediate components.      | Theme, authentication, language, current user.                | 🌍 **Shared Notice Board** |
| **Context Provider** | A React component that **provides shared state** to all its descendant components.                            | To make one source of data available throughout a component tree. | Wrap the part of the app that needs shared data.              | 📡 **Wi-Fi Router**        |
| **Redux**            | An external library that stores **global application state in a centralized store** with predictable updates. | To manage large, complex, frequently changing shared state.       | Large applications (cart, orders, notifications, dashboards). | 🏦 **Central Bank Vault**  |

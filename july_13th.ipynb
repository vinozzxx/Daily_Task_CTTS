JWT = A token format (a type of ID card).

Access Token = The token you send with API requests (often implemented as a JWT).

Refresh Token = Used only to obtain a new access token.

Cookie = A storage mechanism in the browser.

Session Key = A temporary encryption key used by HTTPS/TLS, unrelated to authentication.

Protected Route = Frontend logic that decides whether to show a page based on authentication state.

| Concept               | **What?**                                                     | **Why?**                             | **When to Use?**                         | **Memory Trick**              |
| --------------------- | ------------------------------------------------------------- | ------------------------------------ | ---------------------------------------- | ----------------------------- |
| **GitLens**           | VS Code extension that shows Git history, author, and changes | Understand who changed code and when | While working with Git projects          | 🔍 **Git X-Ray**              |
| **Directory Path**    | Address of a file or folder in a computer                     | Locate files correctly               | Reading files, imports, configuration    | 🗺️ **Home Address**          |
| **JavaScript**        | Dynamic scripting language                                    | Build interactive websites           | Frontend and backend (Node.js)           | 🚗 **Normal Car**             |
| **TypeScript**        | JavaScript + static types                                     | Catch errors before running          | Large, maintainable projects             | 🚙 **Car with GPS**           |
| **Synchronous**       | Tasks run one after another                                   | Simple execution                     | Small or dependent tasks                 | 🚶 **One Person in Line**     |
| **Asynchronous**      | Tasks run without waiting                                     | Faster, non-blocking programs        | API calls, file downloads                | 🍽️ **Order Food, Eat Later** |
| **Single Thread**     | One task at a time                                            | Simpler execution                    | JavaScript runtime                       | 👤 **One Worker**             |
| **Multi Thread**      | Multiple tasks at the same time                               | Better performance                   | Java, Python threading, C#               | 👥 **Many Workers**           |
| **Callback Hell**     | Nested callbacks become hard to read                          | Problem caused by async callbacks    | Older JavaScript code                    | 🪜 **Staircase of Doom**      |
| **ORM**               | Converts database tables into objects                         | Avoid writing raw SQL repeatedly     | Django ORM, SQLAlchemy, Laravel Eloquent | 🗣️ **Translator**            |
| **ClickHouse**        | High-speed analytical database                                | Process huge datasets quickly        | Analytics, dashboards, logs              | ⚡ **Formula 1 Database**      |
| **package.json**      | Node.js project configuration                                 | Manage packages and scripts          | JavaScript/TypeScript projects           | 📦 **Node Project Box**       |
| **composer.json**     | PHP project configuration                                     | Manage PHP packages                  | Laravel, Symfony projects                | 📦 **PHP Project Box**        |
| **API Response**      | Data returned by an API                                       | Send results to client               | Every API request                        | 📬 **Reply Letter**           |
| **CRUD**              | Create, Read, Update, Delete                                  | Basic database operations            | Every application                        | ✏️ **Notebook Actions**       |
| **HTTP Status Codes** | Numbers showing request result                                | Tell client success or failure       | Every API response                       | 🚦 **Traffic Lights**         |
| **CI/CD**             | Automates build, testing, and deployment                      | Faster and safer releases            | Modern software development              | 🤖 **Automatic Factory**      |
| **Coding Standards**  | Rules for writing clean code                                  | Improve readability and teamwork     | Every project                            | 📏 **Rulebook**               |


| Standard                          | What                                    | Why                                  | When to Use              | Memory Trick             |
| --------------------------------- | --------------------------------------- | ------------------------------------ | ------------------------ | ------------------------ |
| **Meaningful Names**              | Use clear variable and function names   | Makes code easy to understand        | Always                   | 🏷️ **Good Names**       |
| **Consistent Formatting**         | Keep spacing and indentation consistent | Improves readability                 | Always                   | 📏 **Neat Code**         |
| **Comments (When Needed)**        | Explain *why*, not obvious *what*       | Helps others understand decisions    | Complex logic            | 💬 **Explain Why**       |
| **DRY** (Don't Repeat Yourself)   | Don't duplicate code                    | Easier maintenance                   | When logic repeats       | 📋 **Write Once**        |
| **KISS** (Keep It Simple, Stupid) | Prefer simple solutions                 | Easier to debug                      | Always                   | 😊 **Keep It Simple**    |
| **SOLID**                         | Design object-oriented code well        | Makes code flexible and maintainable | OOP projects             | 🧱 **Strong Foundation** |
| **Error Handling**                | Handle errors gracefully                | Prevents crashes                     | File, DB, API operations | 🛡️ **Be Prepared**      |
| **Follow Language Style Guide**   | Follow official style rules             | Team consistency                     | Always                   | 📘 **Official Rules**    |
| **Small Functions**               | One function = one job                  | Easier to test and reuse             | Always                   | 🔧 **One Job**           |
| **Version Control**               | Use Git properly                        | Track changes and collaborate        | Every project            | 🌿 **Save History**      |

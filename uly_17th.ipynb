| Topic                 | Remember Like This                          | Example                                                                                           |
| --------------------- | ------------------------------------------- | ------------------------------------------------------------------------------------------------- |
| **Tokenization**      | **Split text into tokens**                  | `"I love AI"` → `["I", "love", "AI"]`                                                             |
| **Stop Word Removal** | **Remove common words with little meaning** | `"I love AI in the world"` → `["love", "AI", "world"]` *(removed: I, in, the)*                    |
| **Stemming**          | **Cut words using simple rules**            | `"playing", "played", "plays"` → `"play"`<br>`"studies"` → `"studi"` *(not a real word)*          |
| **Lemmatization**     | **Convert to dictionary form**              | `"running", "ran", "runs"` → `"run"`<br>`"better"` → `"good"`                                     |
| **Root Word**         | **Original/base word**                      | `"unhappiness"` → `"happy"`<br>`"educational"` → `"educate"`                                      |
| **Embeddings**        | **Convert meaning into numbers (vectors)**  | `"I bought a car"` → `[0.23, -0.51, 0.87, ...]`<br>`"I purchased an automobile"` → Similar vector |

| Normal Form | Rule                                         | Problem Solved                           | Memory Trick               |
| ----------- | -------------------------------------------- | ---------------------------------------- | -------------------------- |
| **1NF**     | One cell = One value                         | Removes multiple values in one column    | 📦 One Box = One Item      |
| **2NF**     | Separate data into correct tables            | Removes duplicate customer data          | 👨 Customer has own table  |
| **3NF**     | Every column depends only on the primary key | Removes indirect/transitive dependencies | 🏷️ Category has own table |

| Normal Form | Think Like This                    | Bank Example                                                      |
| ----------- | ---------------------------------- | ----------------------------------------------------------------- |
| **1NF**     | **One cell = One value**           | Don't store two phone numbers in one cell.                        |
| **2NF**     | **One table = One thing (entity)** | Customer, Branch, Employee, and Account each get their own table. |
| **3NF**     | **Don't repeat facts**             | Store "SBI" once in the Bank table, not in every Branch row.      |
